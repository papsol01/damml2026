```python
import numpy as np
import matplotlib as plt
```

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

C_addition = A + B
C_subtraction = A - B
C_mp = np.matmul(A, B)
C_A_transpose = A.T
C_B_transpose = B.T

print(C_addition)
print(C_subtraction)
print(C_mp)
print(C_A_transpose)
print(C_B_transpose)
```
