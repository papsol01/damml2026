Exercise 01.01. Simple Calculator (15-20 minutes) – Write a Python program that takes
two numbers as input from the user and performs basic arithmetic operations (addition,
subtraction, multiplication, and division).
**Instructions:**
- Prompt the user to enter two numbers.
- Perform the four basic arithmetic operations.
- Display the results.

**Example Output:**
- If the user enters 5 and 3, the program should output:**
- 5 + 3 = 8
- 5 - 3 = 2
- 5 * 3 = 15
- 5 / 3 = 1.67 # Two decimal places

```python
a = int(input("Insert number a: "))
b = int(input("Insert number b: "))

def calc(a, b):
    print(a + b)
    print(a - b)
    print(a * b)
    print(a / b)

calc(a,b)
```

## Exercise 01.02.

Grade Evaluation (15-20 minutes) – Create a Python program that evaluates students’ grades based on their scores. Feel free to use the grading scale of your
preference (e.g., see grading systems by country).
**Instructions:**
- Prompt the user to enter a score between 0 and 100.
- Use conditional statements to assign a grade (A, B, C, D, F ) based on the score.
**Example Output:**
- If the user enters 85, according to the academic grading in the United States, the

**program should output:**
- Grade: B

```python
score = int(input("Insert your grade: "))

def eval(score):
    if(score < 20):
        print("F")
    elif(score > 20 and score < 40):
        print("D")
    elif(score > 40 and score < 60):
        print("C")
    elif(score > 60 and score < 80):
        print("B")
    else: print("A")

eval(score)
```
