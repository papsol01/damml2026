```python
score = int(input("Insert your grade: "))

def evalGrade(score):
    if score >= 90:
        print("A")
    elif score >= 80:
        print("B")
    elif score >= 70:
        print("C")
    elif score >= 60:
        print("D")
    else:
        print("F")

evalGrade(score)
```

## Exercise 01.03.

Manage a List of Students (15-20 minutes) – Create a Python program that manages a list of student names.
The program should allow users to add new names, remove existing ones, and display all names in the list.
 Optionally, you can save the list to a file and load it when the program starts.

**Instructions:**
- Initialize an empty list to store student names.
- Provide options for the user to add a new name, remove a name, or display all names in the list.
- Use a loop to offer these options until the user exits.

**Menu Options:**
- This is an example of the menu options you can provide:
1. Add a name
2. Remove a name
3. Display all names
4. Exit

```python
studentNames = []

print("Menu options:")
print("1 - Add a name")
print("2 - Remove a name")
print("3 - Display all names")
print("Exit")

while True:

    print()
    print("------------------------")
    print("Menu options:")
    print("1 - Add a name")
    print("2 - Remove a name")
    print("3 - Display all names")
    print("4 - Exit")
    print("-----------------------")

    answer = input()

    match answer:
        case "1":
            name = input("Write name: ").lower()
            studentNames.append(name)
            print(f"Added {name} to the list of students")
        case "2":
            name = input("Name to remove ").lower()
            if name in studentNames:
                studentNames.remove(name)
                print()
                print(f"Removed {name} from list")
            else:
                print()
                print("Name not in list")
        case "3":
            print("List of all the students:")
            print(studentNames)
        case "4":
            print()
            print("Exited program")
            break
        case _:
            print("Invalid option, try again")
```

## Exercise 01.04.

Simple Contact List with Dictionaries (15-20 minutes) – Develop a Python program that simulates a simple contact list using dictionaries.
The program should allow users to add new contacts, remove existing contacts, search for a contact by name, and display all contacts.
 Optionally, you can save the dictionary to a text file and load it when the program starts.

**Instructions:**
- Use a dictionary to store contacts; the keys are the contacts' names, and the values are their phone numbers.
- Provide options for the user to add a new contact, remove a contact, or search for one.
- Display the contact list in a readable format.
**Example Output:**
- If the user wants to display all contacts, the program should output:
1. Contacts:
2. Alice: +45 12 34 56 78
3. Bob: +45 98 76 54 32
4. Charlie: +45 23 45 67 89


```python
class ContactList:
    def __init__(self):
        self.phoneBook = dict([
            ("Alice", "+45 12 34 56 78"),
            ("Bob", "+45 98 76 54 32"),
            ("Charlie", "+45 23 45 67 89")
        ])

    def addContact(self, name, number):
        self.phoneBook[name] = number

    def removeContact(self, name):
        if name in self.phoneBook:
            self.phoneBook.pop(name)

    def displayContacts(self):
        print("Contacts:")
        for name, number in self.phoneBook.items():
            print(f" {name}: {number}")

cl = ContactList()
cl.removeContact("Bob")
cl.displayContacts()
```

Load the dataset into a Pandas DataFrame and calculate each student's basic statistics (e.g., average grade, standard deviation, minimum, maximum).

**Instructions:**

Were using jupyter


```python
import pandas as pd

df = pd.read_csv("./exercises/data/student_grades.csv")
df.head()

df.isnull()

subjects = df.columns.drop("Student")
df[subjects] = df[subjects].fillna((df[subjects].mean()))
df.isnull().sum()

df["Average"] = df[subjects].mean(axis=1).round(2)

df[["Student", "Average"]]

threshold = 85
above = df[df["Average"] > threshold]
above[["Student", "Average"]]
```

# Exercise 01.06 — Sales Data Analysis

Analyze a small sales dataset. Load it into a Pandas `DataFrame`, calculate
the total sales for each product, determine the day with the highest sales,
and generate summary statistics.

## Learning Objectives
- Load a CSV file into a DataFrame
- Preview data with `head()`
- Group and aggregate data with `groupby()`
- Find the maximum with `idxmax()` and `max()`
- Compute summary statistics (mean, median, mode)


```python
import pandas as pd

df = pd.read_csv("./exercises/data/sales_data.csv")
df.head()
print({df.shape})
```

```python
sales_per_product = df.groupby("Product")["Sales Amount"].sum()
sales_per_product
```

```python
sales_per_day = df.groupby("Date")["Sales Amount"].sum()
sales_per_day
```

```python
best_day = sales_per_day.idxmax()
best_total = sales_per_day.max()

print(f"Day with the highest sales: {best_day} ({best_total})")
```

```python
print("Mean: ", df["Sales Amount"].mean())
print("Median: ", df["Sales Amount"].median())
print("Mode: ", df["Sales Amount"].mode()[0])
```
