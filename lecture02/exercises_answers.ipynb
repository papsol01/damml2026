```python
import matplotlib
matplotlib.use("module://matplotlib_inline.backend_inline")
import random
import matplotlib.pyplot as plt
import numpy as np
from pandas import DataFrame
from scipy.stats import norm
```

```python
def create_drunken(num: int) -> list:
    """Create a list of 'num'(ber) drunken people.

    Args:
        num (int): Number of drunken people.

    Returns:
        list: A list of drunken people.
    """
    s = []

    for _ in range(num):
        s.append(0)

    return s


def single_step(rand: float) -> int:
    """Make a drunken person make one step.

    Args:
        rand (float): Random number to decide the direction of the step.

    Returns:
        int: The step to be taken.
    """
    step = 0

    if rand < 0 or rand > 1:
        print("Random variable is not in the ]0,1[ interval.")
    elif rand < 0.5:
        step = step - 1
    else:
        step = step + 1

    return step


def step(drunken: list) -> list:
    """Make a set of drunken people take a (mis)step.

    Args:
        drunken (list): A list of drunken people.

    Returns:
        list: A list of drunken people after the step.
    """
    ret = []

    for dr in drunken:
        move = random.uniform(0, 1)
        dr = dr + single_step(move)
        ret.append(dr)

    return ret

"""First check point achieved.

We now plot a little bit. The idea is:
    1) One create a list of drunken people.
    2) Make them walk a bit.
    3) Make an block plot and look at it.
"""

def random_walk(drunken: list, num: int) -> list:
    """Random walk with N steps.

    Args:
        drunken (list): A list of drunken people.
        num (int): Number of steps.

    Returns:
        list: A list of drunken people after the steps.
    """
    ret = drunken

    for _ in range(num):
        ret = step(ret)

    return ret


def plot_drunken(drunken: list, bins: int):
    """Plot the drunken people in a histogram.

    Args:
        drunken (list): A list of drunken people.
        bins (int): Number of bins.
    """
    plt.hist(drunken, bins)
    plt.show()

"""Second check point achieved.

We now compare histogram and try to check how Gauss is this distribution:
   1) Make a lot of drunken people walk.
   2) Make a Gaussian fit.
   3) Make a block plot.
   4) Plot the Gauss fit over it.
   5) Calculate the average and standard deviation "by hand".
   6) Compare it with obtained by the fit.
"""

def export_data(data: list, filename: str):
    """Export the data to a CSV file.

    Args:
        data (list): List of data.
        filename (str): Name of the file.
    """
    df = DataFrame({ "Druken People final position": data })
    df.to_csv(filename)


def normal_drunkens(drunken: list) -> tuple:
    """Make a Gaussian fit of the drunken people.

    Args:
        drunken (list): A list of drunken people.

    Returns:
        tuple: Average and standard deviation.
    """
    mu, std = norm.fit(drunken)
    print(f"Median = {mu}\nStandard Deviation = {std}")
    return mu, std


def normal_fit_plot(drunken: list, bins: int, mu: np.float64, std: np.float64):
    """Plot the histogram and the Gaussian fit.

    Args:
        drunken (list): A list of drunken people.
        bins (int): Number of bins.
        mu (np.float64): Average.
        std (np.float64): Standard deviation.
    """
    plt.hist(drunken, bins, density=True)

    a, b = plt.xlim()
    x = np.linspace(a, b, 1000)
    fit = norm.pdf(x, mu, std)

    plt.plot(x, fit, "k", linewidth=2)
    plt.show()
```
